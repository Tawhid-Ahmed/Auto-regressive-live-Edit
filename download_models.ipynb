```python
# Configuration - Model paths
import os

# Base path for models (matches GLOBAL.py)
BASE_MODEL_PATH = r"E:\MSC\MSc thesis project\LiveEdit\models"

# Model configurations
MODELS = {
    'blip2-opt-2.7b': {
        'hf_name': 'Salesforce/blip2-opt-2.7b',
        'local_path': os.path.join(BASE_MODEL_PATH, 'blip2-opt-2.7b')
    },
    'llava-v1.5-7b': {
        'hf_name': 'llava-hf/llava-1.5-7b-hf',
        'local_path': os.path.join(BASE_MODEL_PATH, 'llava-v1.5-7b-hf')
    },
    'minigpt-4-vicuna-7b': {
        'hf_name': None,  # MiniGPT-4 requires special setup
        'local_path': os.path.join(BASE_MODEL_PATH, 'minigpt-4-vicuna-7b')
    }
}

# Create base directory
os.makedirs(BASE_MODEL_PATH, exist_ok=True)

print("Configuration loaded!")
print(f"Models will be saved to: {BASE_MODEL_PATH}")
```

## 1. Download BLIP2-OPT-2.7B Model

This model is approximately 5-6 GB. The download may take 10-30 minutes depending on your internet speed.

```python
# Download BLIP2-OPT-2.7B
from transformers import Blip2ForConditionalGeneration, Blip2Processor
import os

model_name = 'blip2-opt-2.7b'
model_info = MODELS[model_name]
local_path = model_info['local_path']
hf_name = model_info['hf_name']

print(f"Downloading {model_name}...")
print(f"From: {hf_name}")
print(f"To: {local_path}")
print("This may take 10-30 minutes depending on your internet speed...\n")

# Create directory
os.makedirs(local_path, exist_ok=True)

try:
    # Download model
    print("Downloading model weights...")
    model = Blip2ForConditionalGeneration.from_pretrained(hf_name)
    model.save_pretrained(local_path)
    print("✓ Model weights downloaded")
    
    # Download processor
    print("Downloading processor...")
    processor = Blip2Processor.from_pretrained(hf_name)
    processor.save_pretrained(local_path)
    print("✓ Processor downloaded")
    
    print(f"\n✅ {model_name} successfully downloaded to: {local_path}")
    
except Exception as e:
    print(f"\n❌ Error downloading {model_name}: {str(e)}")
    print("Make sure you have:")
    print("1. Sufficient disk space")
    print("2. Stable internet connection")
    print("3. HuggingFace account (login with: huggingface-cli login)")
```

```python
# Download LLaVA v1.5 7B
from transformers import LlavaForConditionalGeneration, LlavaProcessor
import os

model_name = 'llava-v1.5-7b'
model_info = MODELS[model_name]
local_path = model_info['local_path']
hf_name = model_info['hf_name']

print(f"Downloading {model_name}...")
print(f"From: {hf_name}")
print(f"To: {local_path}")
print("This may take 20-40 minutes depending on your internet speed...\n")

# Create directory
os.makedirs(local_path, exist_ok=True)

try:
    # Download model
    print("Downloading model weights...")
    model = LlavaForConditionalGeneration.from_pretrained(hf_name)
    model.save_pretrained(local_path)
    print("✓ Model weights downloaded")
    
    # Download processor
    print("Downloading processor...")
    processor = LlavaProcessor.from_pretrained(hf_name)
    processor.save_pretrained(local_path)
    print("✓ Processor downloaded")
    
    print(f"\n✅ {model_name} successfully downloaded to: {local_path}")
    
except Exception as e:
    print(f"\n❌ Error downloading {model_name}: {str(e)}")
    print("Make sure you have:")
    print("1. Sufficient disk space")
    print("2. Stable internet connection")
    print("3. HuggingFace account (login with: huggingface-cli login)")
```

## Verification

Check if all models are downloaded correctly.

```python
# Verify downloaded models
import os

print("Verifying downloaded models...\n")
print("=" * 60)

all_ok = True

# Check BLIP2
blip2_path = MODELS['blip2-opt-2.7b']['local_path']
if os.path.exists(blip2_path):
    files = os.listdir(blip2_path)
    if len(files) > 0:
        print(f"✅ BLIP2-OPT-2.7B: Found at {blip2_path}")
        print(f"   Files: {len(files)} files")
    else:
        print(f"⚠️  BLIP2-OPT-2.7B: Directory exists but is empty")
        all_ok = False
else:
    print(f"❌ BLIP2-OPT-2.7B: Not found at {blip2_path}")
    all_ok = False

print()

# Check LLaVA
llava_path = MODELS['llava-v1.5-7b']['local_path']
if os.path.exists(llava_path):
    files = os.listdir(llava_path)
    if len(files) > 0:
        print(f"✅ LLaVA v1.5 7B: Found at {llava_path}")
        print(f"   Files: {len(files)} files")
    else:
        print(f"⚠️  LLaVA v1.5 7B: Directory exists but is empty")
        all_ok = False
else:
    print(f"❌ LLaVA v1.5 7B: Not found at {llava_path}")
    all_ok = False

print()

# Check MiniGPT-4
minigpt_path = MODELS['minigpt-4-vicuna-7b']['local_path']
if os.path.exists(minigpt_path):
    files = os.listdir(minigpt_path)
    if len(files) > 0:
        print(f"✅ MiniGPT-4: Found at {minigpt_path}")
        print(f"   Files: {len(files)} files")
        # Check for required config files
        has_config = any('config' in f.lower() for f in files)
        if has_config:
            print(f"   ✓ Config files found")
        else:
            print(f"   ⚠️  Config files may be missing")
    else:
        print(f"⚠️  MiniGPT-4: Directory exists but is empty")
        all_ok = False
else:
    print(f"❌ MiniGPT-4: Not found at {minigpt_path}")
    all_ok = False

print()
print("=" * 60)

if all_ok:
    print("✅ All models appear to be downloaded!")
else:
    print("⚠️  Some models are missing. Please download them using the cells above.")
```

## Troubleshooting

### If you encounter authentication errors:
1. Create a HuggingFace account at https://huggingface.co
2. Login using: `huggingface-cli login`
3. Accept any model licenses if required

### If downloads are slow or fail:
- Check your internet connection
- Try downloading during off-peak hours
- Some models may require Git LFS: `pip install git-lfs && git lfs install`

### If you run out of disk space:
- Models require ~20-30 GB total
- Free up space or download to an external drive
- Update paths in `utils/GLOBAL.py` accordingly
